```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

final_tails = []

for i in range(100):
    tails = [0]

    for j in range(10):
        coin = np.random.randint(0, 2)
        tails.append(tails[-1] + coin)

    final_tails.append(tails[-1])

print(final_tails)
```

```python
plt.hist(final_tails, bins=10)
plt.show()
```
